# T-test

###t-test — kratka obnova z vsemi osnovami

###Kaj je in zakaj ga uporabljamo

**t-test** preverja, ali je opažena **razlika v povprečjih** (ali odklon povprečja od ciljne vrednosti) **večja od naključnega šuma** (razpršenosti podatkov glede na velikost vzorca).
Uporabimo ga, ko je **Y numerična** mera (čas, cena, dolžina …) in želimo primerjati:

* **eno povprečje** z znano referenco (enovzorčni),
* **dve povprečji** neodvisnih skupin (A vs. B),
* **povprečje razlik** pri **pred–po** na istih enotah (parni).

> Dvorepni test odgovarja: “ali je **kakršnakoli** razlika?”
> Enorepni: “ali je razlika **v izbrano smer**?”

---

###Kaj v resnici “dela”

Izračuna razmerje

$$
t=\frac{\text{ocenjena razlika}}{\text{standardna napaka te razlike}}
$$

in ga primerja s **t-porazdelitvijo** (glede na **stopinje prostosti**, df). Velik $|t|$ ⇒ majhna p-vrednost ⇒ malo verjetno, da razlika izhaja samo iz šuma.

---

###Enačbe (3 najpogostejši primeri)

####(1) Enovzorčni t-test

Preverimo, ali se povprečje razlikuje od $\mu_0$.

$$
t=\frac{\bar x-\mu_0}{s/\sqrt{n}},\quad \text{df}=n-1
$$

* $\bar x$ — povprečje vzorca; $s$ — **vzročni** SD; $n$ — velikost vzorca.
* **SE** povprečja je $s/\sqrt{n}$.

####(2) Dvo-vzorčni t-test (neodvisni skupini A, B)

**Welch** (varna privzeta izbira, ne predpostavlja enakih varianc):

$$
t=\frac{\bar x_A-\bar x_B}{\sqrt{\frac{s_A^2}{n_A}+\frac{s_B^2}{n_B}}},\quad 
\text{df}\approx
\frac{\left(\frac{s_A^2}{n_A}+\frac{s_B^2}{n_B}\right)^2}
{\frac{(s_A^2/n_A)^2}{n_A-1}+\frac{(s_B^2/n_B)^2}{n_B-1}}
$$

**Pooled** (le če sta varianci \~ enaki):

$$
s_p=\sqrt{\frac{(n_A-1)s_A^2+(n_B-1)s_B^2}{n_A+n_B-2}},\quad
t=\frac{\bar x_A-\bar x_B}{s_p\sqrt{\frac{1}{n_A}+\frac{1}{n_B}}},\ 
\text{df}=n_A+n_B-2
$$

####(3) Parni (paired) t-test

Na istih enotah izračunamo razlike $d_i=x_{i,\text{po}}-x_{i,\text{pred}}$, nato

$$
t=\frac{\bar d}{s_d/\sqrt{n}},\quad \text{df}=n-1
$$

---

###Razlaga členov v enačbah

* $\bar x$, $\bar x_A$, $\bar x_B$, $\bar d$: **ocenjena razlika** (povprečje ali razlika povprečij).
* $s, s_A, s_B, s_d$: **razpršenost** (standardni odklon) — “šum”.
* $\sqrt{\frac{s_A^2}{n_A}+\frac{s_B^2}{n_B}}$ ali $s/\sqrt{n}$: **standardna napaka (SE)** ocene — večji vzorec ⇒ manjša SE.
* **df (stopinje prostosti)**: določa obliko t-porazdelitve za p-vrednost in interval zaupanja.

---

###Predpostavke (in robustne alternative)

* Y je **približno normalno** porazdeljena znotraj skupin (pri majhnem $n$); pri večjem $n$ pomaga CLT.
* Opazovanja so **neodvisna** (razen pri paired, kjer so pari povezani).
* Za “pooled” varianci ≈ enaki; če dvom, uporabi **Welch**.
* Če močna odstopanja/outlierji/ordinalni podatki: **Mann–Whitney U** (namesto dvo-vzorčnega), **Wilcoxon signed-rank** (namesto paired).

---

###Poročanje rezultatov

Vedno podaj:

* $\Delta$ (razlika povprečij) in **95 % CI** za $\Delta$,
* $t(\text{df})$ in **p-vrednost** (dvorepna, razen če je upravičeno enorepna),
* **velikost učinka** (npr. **Cohenov d**: $d=\frac{\bar x_A-\bar x_B}{s_p}$).

---

###Kratek, konkreten primer (dvo-vzorčni Welch t-test)

**Vprašanje:** Ali je povprečni **čas obdelave** drugačen med izmeno **A** in **B**?
**Podatki (min):**
A = \[6.1, 5.9, 6.3, 6.0, 6.2] → $\bar x_A=6.10$, $s_A^2=0.025$
B = \[5.5, 5.8, 5.9, 5.7, 5.6] → $\bar x_B=5.70$, $s_B^2=0.025$

1. **Razlika povprečij:** $\Delta=\bar x_A-\bar x_B=0.40$ min
2. **SE razlike:** $\sqrt{0.025/5+0.025/5}=\sqrt{0.01}=0.10$
3. **t-statistika:** $t=0.40/0.10=4.00$
4. **df (Welch):** $\approx 8$
5. **p (dvorepna):** $\approx 0.003$ ⇒ **značilna razlika**
6. **95 % CI za $\Delta$:** $0.40 \pm t_{0.975,8}\cdot 0.10 = 0.40 \pm 2.306\cdot 0.10$
   ⇒ **\[0.17, 0.63] min**
7. **Cohenov d (pooled):** $s_p=\sqrt{0.025}=0.158$ ⇒ $d=0.40/0.158\approx 2.53$ (zelo velik učinek)

**Interpretacija (1–2 stavka):**
“Izmena A ima v povprečju za \~0.40 min daljši čas obdelave kot B; razlika je statistično značilna, $t(8)=4.00$, $p\approx0.003$, 95 % CI \[0.17, 0.63]. Učinek je velik (d≈2.5).”

---

####Namig za prakso

* Privzeto uporabi **Welch t-test** (robusten na neenake variance).
* Če smer vnaprej res upravičiš (npr. “po izboljšavi je **manj**”), lahko uporabiš **enorepni** test; sicer **dvorepni**.
* Za $k>2$ skupin pojdi na **ANOVO** (ali regresijo z dummy spremenljivkami).

---

###(1) Enovzorčni t-test

**Primer:** povprečni čas obdelave naročila naj bi bil **6.5 min**. Imaš vzorec 5 meritev:
`[5, 6, 7, 6, 6]` (min). Ali je povprečje **različno** od 6.5?

###Koraki

**K1) Postavi hipoteze in izberi test**

* $H_0:\ \mu = 6.5$
* $H_1:\ \mu \neq 6.5$ (dvorepni test)
  **Zakaj:** zanima nas **kakršnakoli** razlika od cilja.

**K2) Izračunaj osnovne statistike**

* Velikost vzorca: $n=5$
* Povprečje: $\bar x=\frac{5+6+7+6+6}{5}=6.0$
* Odkloni: $(-1,0,1,0,0)$ ⇒ vsota kvadratov $=2$
* Vzročni SD: $s=\sqrt{\frac{2}{n-1}}=\sqrt{\frac{2}{4}}=\sqrt{0.5}=0.7071$
* Standardna napaka (SE): $\text{SE}=s/\sqrt{n}=0.7071/\sqrt{5}=0.3162$
  **Zakaj:** SE meri “**šum**” povprečja glede na razpršenost in velikost vzorca.

**K3) Testna statistika**

$$
t=\frac{\bar x-\mu_0}{\text{SE}}=\frac{6.0-6.5}{0.3162}=-1.581
$$

**Zakaj:** razmerje “učinek / šum”.

**K4) Stopinje prostosti**

* $\text{df}=n-1=4$.
  **Zakaj:** t-porazdelitev za vzorčni SD potrebuje df.

**K5) p-vrednost in 95 % CI**

* Dvorepna $p \approx 0.19$ (za $|t|=1.581$ in df=4).
* 95 % CI: $\bar x \pm t_{0.975,4}\cdot \text{SE}=6.0 \pm 2.776\cdot 0.3162$
  ⇒ $6.0 \pm 0.878$ ⇒ **\[5.122, 6.878]**
  **Zakaj:** p pove “ali je razlika večja od šuma”, CI pokaže **velikost in negotovost**.

**K6) Interpretacija**

* Razlike **ni** statistično značilne (p≈0.19). Povprečje je \~6.0 min; 95 % CI \[5.12, 6.88].

---

##(2) Dvo-vzorčni t-test (neodvisni skupini A in B)

**Primer:** primerjamo povprečni čas dveh izmen.

* A: `[6.1, 5.9, 6.3, 6.0, 6.2]`
* B: `[5.5, 5.8, 5.9, 5.7, 5.6]`

###Koraki

**K1) Hipoteze**

* $H_0:\ \mu_A=\mu_B$
* $H_1:\ \mu_A\neq\mu_B$ (dvorepni)
  **Zakaj:** zanima nas, ali se skupini razlikujeta.

**K2) Osnovne statistike**

* $\bar x_A=6.10$, $\bar x_B=5.70$ ⇒ razlika $\Delta=0.40$
* Vzročni varianci: $s_A^2=s_B^2=0.025$ (izračunano iz podatkov)
* $n_A=n_B=5$

**K3) Izbira različice t-testa**

* Uporabimo **Welch** (varna privzeta, ne zahteva enakih varianc).
  **Zakaj:** robustnost; tudi če sta varianci enaki, rezultat je skoraj isti.

**K4) Testna statistika (Welch)**

$$
\text{SE}=\sqrt{\frac{s_A^2}{n_A}+\frac{s_B^2}{n_B}}=\sqrt{0.025/5+0.025/5}=\sqrt{0.01}=0.10
$$

$$
t=\frac{\bar x_A-\bar x_B}{\text{SE}}=\frac{0.40}{0.10}=4.00
$$

**Zakaj:** SE meri šum razlike med povprečji.

**K5) Stopinje prostosti (Welch)**

$$
\text{df}\approx
\frac{(0.025/5+0.025/5)^2}{\frac{(0.025/5)^2}{5-1}+\frac{(0.025/5)^2}{5-1}}
=\frac{0.01^2}{\frac{0.005^2}{4}+\frac{0.005^2}{4}}
=\frac{0.0001}{0.0000125}=8
$$

**Zakaj:** Welchova formula za neenake variance/velikosti.

**K6) p-vrednost in 95 % CI**

* $p$ (dvorepno, df=8, $t=4.00$) $\approx 0.003$
* 95 % CI za $\Delta$: $0.40 \pm t_{0.975,8}\cdot 0.10=0.40 \pm 2.306\cdot 0.10$
  ⇒ **\[0.17, 0.63] min**
  **Zakaj:** CI pove, koliko je A počasnejša od B v povprečju.

**K7) Interpretacija**

* Izmena **A** je počasnejša za \~0.40 min (95 % CI \[0.17, 0.63]); razlika je **statistično značilna** (p≈0.003).

---

##(3) Parni (paired) t-test

**Primer:** meriš **isti** sklop 6 naročil **pred** in **po** izboljšavi (parjene meritve).

* **Pred:** `[6.0, 6.1, 6.2, 5.9, 6.0, 6.3]`
* **Po:**   `[5.0, 5.3, 5.6, 5.0, 5.3, 5.3]`

Izračunaj razlike $d_i=\text{po}-\text{pred}$:
$[-1.0,\ -0.8,\ -0.6,\ -0.9,\ -0.7,\ -1.0]$

###Koraki

**K1) Hipoteze**

* $H_0:\ \mu_d=0$ (ni povprečne spremembe)
* $H_1:\ \mu_d\neq 0$ (dvorepno) ali $<0$ (enorepno, če te zanima “po je manj”)
  **Zakaj:** testiramo **povprečje razlik** na istih enotah.

**K2) Osnovne statistike za razlike**

* $n=6$
* $\bar d=\frac{-1.0-0.8-0.6-0.9-0.7-1.0}{6}=-0.8333$
* Vsota kvadratov odklonov od $\bar d$ = $0.1333$
* $s_d=\sqrt{\frac{0.1333}{n-1}}=\sqrt{0.0267}=0.1633$
* $\text{SE} = s_d/\sqrt{n}=0.1633/\sqrt{6}=0.0667$
  **Zakaj:** vse računamo **na razlikah**, ker so pari povezani.

**K3) Testna statistika**

$$
t=\frac{\bar d}{\text{SE}}=\frac{-0.8333}{0.0667}=-12.50
$$

**K4) Stopinje prostosti**

* $\text{df}=n-1=5$.

**K5) p-vrednost in 95 % CI**

* $p$ (dvorepno) je **zelo majhen** ($<0.0001$).
* 95 % CI: $\bar d \pm t_{0.975,5}\cdot \text{SE}=-0.8333 \pm 2.571\cdot 0.0667$
  ⇒ **\[-1.005, −0.662]**
  **Zakaj:** CI pove, za koliko se je povprečje zmanjšalo po izboljšavi.

**K6) Interpretacija**

* Po izboljšavi je čas **nižji** za \~0.83 min (95 % CI \[−1.01, −0.66]); razlika je **močno značilna**.

---

###Hitri povzetek razlik med tremi testi

* **Enovzorčni:** eno povprečje proti znani vrednosti.
* **Dvo-vzorčni (neodvisni):** dve **neodvisni** skupini.
* **Parni:** isti primerki **pred–po** (ali ujemajoči pari).

V vseh treh primerih je logika ista:
**t = (ocenjena razlika) / (standardna napaka)**, potem **df**, **p-vrednost**, **CI** in **interpretacija** v jeziku problema.


# ANOVAŽ

Super, gremo čisto jasno in praktično.

## ANOVA — kaj je, enačbe, primer, razlika s t-testom

## Kaj je ANOVA?

**ANOVA (Analysis of Variance / analiza variance)** preverja, ali se **povprečja** numerične spremenljivke **razlikujejo med več kot dvema skupinama** (k ≥ 2).
Ideja: razdeli skupno varianco na:

* **medskupinsko** (razlike med povprečji skupin) in
* **znotrajskupinsko** (šum znotraj vsake skupine).

Če je “medskupinska” varianca veliko večja od “znotrajskupinske”, sklepamo, da se **vsaj eno** skupinsko povprečje razlikuje.

## Kdaj uporabimo ANOVO?

* Y **numerična**, X **kategorna s k ≥ 2** nivoji (npr. trije stroji, tri izmene, trije načini postopka).
* Za **k=2** se ANOVA skrči na **t-test** (glej spodaj *“kdaj je enaka t-testu”*).

## Enačbe (ena-faktorska ANOVA, klasična različica)

Naj bo k skupin, skupina j ima $n_j$ opazovanj, skupno $N=\sum_j n_j$.

* Skupinsko povprečje: $\bar y_j$
* Skupno (grand) povprečje: $\bar y=\frac{1}{N}\sum_{j}\sum_{i} y_{ij}$

**Razcep sum kvadratov:**

$$
\underbrace{\text{SST}}_{\text{skupna}}=\sum_{j=1}^k\sum_{i=1}^{n_j}(y_{ij}-\bar y)^2
=\underbrace{\text{SSB}}_{\text{med skupinami}}+\underbrace{\text{SSW}}_{\text{znotraj skupin}}
$$

kjer

$$
\text{SSB}=\sum_{j=1}^k n_j(\bar y_j-\bar y)^2,\qquad
\text{SSW}=\sum_{j=1}^k\sum_{i=1}^{n_j}(y_{ij}-\bar y_j)^2
$$

**Povprečni kvadrati in F-statistika:**

$$
\text{MSB}=\frac{\text{SSB}}{k-1},\qquad
\text{MSW}=\frac{\text{SSW}}{N-k},\qquad
F=\frac{\text{MSB}}{\text{MSW}}
$$

* Stopinje prostosti: $\text{df}_1=k-1$ (med), $\text{df}_2=N-k$ (znotraj).
* Če je $F$ “velik” (p-majhna), vsaj eno povprečje odstopa.

**Efektna mera (priporočljivo):**
$\eta^2=\text{SSB}/\text{SST}$ (delež razložene variance).
Bolje pristranostno-korigirano: $\omega^2$.

---

## Primer (korak-za-korakom račun)

**Vprašanje:** Ali se povprečni **čas obdelave** razlikuje med tremi stroji (A, B, C)?

**Podatki (min):**
A = \[6.1, 5.9, 6.2]
B = \[5.6, 5.7, 5.8]
C = \[6.3, 6.4, 6.5]

### 1) Osnovne vrednosti

* $\bar y_A=6.067$, $\bar y_B=5.700$, $\bar y_C=6.400$
* Skupno povprečje $\bar y=6.056$
* $k=3$, $N=9$

**Zakaj:** potrebujemo skupinska in skupno povprečje za razcep variance.

### 2) Izračun SSB in SSW

* $\text{SSB}=\sum n_j(\bar y_j-\bar y)^2=0.7356$
* $\text{SSW}=\sum\sum (y_{ij}-\bar y_j)^2=0.0867$
* $\text{SST}=0.8222$ (in res $\text{SSB}+\text{SSW}=\text{SST}$)

**Zakaj:** SSB meri “medskupinsko” razliko (signal), SSW “znotrajskupinski” šum.

### 3) MSB, MSW, F

* $\text{MSB}=\text{SSB}/(k-1)=0.3678$
* $\text{MSW}=\text{SSW}/(N-k)=0.01444$
* $F=\text{MSB}/\text{MSW}=25.462$
* df$_1=2$, df$_2=6$
* $p\approx 0.00117$ (dvorepno za F)

**Zakaj:** F je razmerje “signal/šum”. Velik F ⇒ razlike med povprečji so večje od šuma.

### 4) Efektna mera

* $\eta^2=\text{SSB}/\text{SST}=0.7356/0.8222\approx 0.895$
  (≈ 89.5 % variance v času obdelave pojasni “kateri stroj”.)

### 5) Interpretacija

“Razlike med stroji so **statistično značilne**: $F(2,6)=25.46$, $p\approx 0.0012$.
Efekt je zelo velik ($\eta^2\approx0.90$).”

**Opomba:** ANOVA pove, **da obstaja vsaj ena razlika**.
Če želiš vedeti **kateri pari** se razlikujejo, naredi **post-hoc** (npr. **Tukey HSD**).

---

## Kako se razlikuje od t-testa in kdaj je enaka?

* **Razlika:**

  * **t-test** primerja **2 skupini** (A vs B).
  * **ANOVA** lahko primerja **k skupin** hkrati (k ≥ 2) in kontrolira skupno napako tipa I.
* **Enakost s t-testom (k=2):**

  * Pri **ena-faktorski ANOVI** z **2 skupinama** velja $\;F=t^2\;$.
  * To je enako **klasičnemu (pooled) dvo-vzorčnemu t-testu** (predpostavka **enakih varianc**).
  * Če varianci nista enaki, uporabiš **Welch ANOVO** (analog **Welch t-testa**).

---

## Predpostavke (in kaj, če ne držijo)

* **Neodvisnost** opazovanj (ključna).
* **Približna normalnost** Y v vsaki skupini (pri majhnem n; pri večjem pomaga CLT).
* **Homoskedastičnost** (približno enake variance v skupinah) za klasično ANOVO.

  * Če ne drži: **Welch ANOVA** (robustna na neenake variance in velikosti vzorcev).
* **Outlierji** lahko izkrivijo rezultat ⇒ preveri boxplote/diagnostiko.

---

## Kratek “recept” za uporabo v praksi

1. **Definiraj Y** (numerična mera) in **X** (kategorialna z ≥2 nivoji).
2. Izračunaj **SSB, SSW, MSB, MSW, F**, **p** in **$\eta^2$/$\omega^2$**.
3. Če $p<\alpha$ ⇒ nadaljuj s **post-hoc** pari (npr. **Tukey HSD**).
4. Poročaj: $F(\text{df}_1,\text{df}_2)=\dots$, $p=\dots$, $\eta^2$ in povzetek, kdo je višji/nižji (iz povprečij).

---

Če želiš, lahko na tvojih konkretnih stolpcih (Y + X) izračunam **klasično ANOVO + Tukey HSD** in **Welch ANOVO**, ter ob tem pokažem, kako je pri k=2 res $F=t^2$.
